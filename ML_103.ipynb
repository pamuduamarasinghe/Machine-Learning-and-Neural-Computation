# Car Dataset analaysis

```python
import pandas as pd
import numpy as np





# b. 
column_names = [
    'symboling', 'normalized-losses', 'make', 'fuel-type', 'aspiration', 
    'num-of-doors', 'body-style', 'drive-wheels', 'engine-location', 
    'wheel-base', 'length', 'width', 'height', 'curb-weight', 'engine-type',
    'num-of-cylinders', 'engine-size', 'fuel-system', 'bore', 'stroke',
    'compression-ratio', 'horsepower', 'peak-rpm', 'city-mpg', 'highway-mpg', 'price'
]

# Load dataset
df = pd.read_csv('Car_dataset.csv', header=None, names=column_names)

print("DataFrame created successfully!")
print(f"Dataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

# Data preprocessing

```python
print("\nMissing values in each column:")
missing_values = df.isnull().sum()
print(missing_values)
```

```python
duplicate_rows = df.duplicated().sum()
unique_rows = df.drop_duplicates().shape[0]
print(f"\nNumber of duplicate rows: {duplicate_rows}")
print(f"Number of unique rows: {unique_rows}")
```

```python
if duplicate_rows > 0:
    print("\nDuplicated rows:")
    print(df[df.duplicated(keep=False)])
else:
    print("\nNo duplicated rows found.")
```

```python
df_new = df.drop_duplicates()
print(f"\nAfter removing duplicates - New DataFrame shape: {df_new.shape}")

df_no_missing = df_new.dropna()
print(f"\nAfter removing all missing values - Remaining rows: {df_no_missing.shape[0]}")


print(f"Current removal: {df_new.shape[0] - df_no_missing.shape[0]} rows removed " +
      f"({((df_new.shape[0] - df_no_missing.shape[0])/df_new.shape[0]*100):.1f}% of data)")
```

```python
numerical_cols = df_new.select_dtypes(include=['float64', 'int64']).columns
categorical_cols = df_new.select_dtypes(include=['object', 'string']).columns


df_mean_filled = df_new.copy()

for col in numerical_cols:
    df_mean_filled[col] = df_new[col].fillna(df_new[col].mean())

for col in categorical_cols:
    df_mean_filled[col] = df_new[col].fillna(df_new[col].mode()[0])

print(f"\nAfter mean/mode imputation - Missing values: {df_mean_filled.isnull().sum().sum()}")


df_median_filled = df_new.copy()

for col in numerical_cols:
    df_median_filled[col] = df_new[col].fillna(df_new[col].median())

for col in categorical_cols:
    df_median_filled[col] = df_new[col].fillna(df_new[col].mode()[0])

print(f"After median/mode imputation - Missing values: {df_median_filled.isnull().sum().sum()}")


```

hot encoding data

```python
print("\nj. One-hot encoding for categorical columns:")
df_encoded = pd.get_dummies(df_mean_filled, columns=categorical_cols)
print(f"After one-hot encoding - Number of columns: {df_encoded.shape[1]}")
print(f"New columns created: {[col for col in df_encoded.columns if any(x in col for x in categorical_cols)]}")
```

# Statistical Analysis

```python
df_stats = df_mean_filled

# only numerical columns 
numerical_cols_for_stats = [col for col in numerical_cols if col in df_stats.columns]

print("\nStatistical parameters for each numerical column:\n")

stats_summary = pd.DataFrame(index=numerical_cols_for_stats)

for col in numerical_cols_for_stats:
    if df_stats[col].dtype in ['int64', 'float64']:

        stats_summary.loc[col, 'Mean'] = df_stats[col].mean()
        

        stats_summary.loc[col, 'Median'] = df_stats[col].median()
        

        mode_values = df_stats[col].mode()
        stats_summary.loc[col, 'Mode'] = mode_values.iloc[0] if len(mode_values) > 0 else 'Multiple'
        

        stats_summary.loc[col, 'Variance'] = df_stats[col].var()
        

        stats_summary.loc[col, 'Standard Deviation'] = df_stats[col].std()

        stats_summary.loc[col, 'Minimum'] = df_stats[col].min()
        

        stats_summary.loc[col, 'Maximum'] = df_stats[col].max()

# Summary
print(stats_summary.round(2))

```
